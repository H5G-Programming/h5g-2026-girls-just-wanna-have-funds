## 5) Dictionaries
Dictionaries store key-value pairs.
The key always comes first followed by the value. 
The pattern looks like
 "key": value

Each dictionary is surrounded by {}
You can either copy them or use shift + option + paranthesis to create them on Mac


```python
user = {
    "id": 1, 
    "name": "Maya", 
    "age": 14
}

print(user)

print(user["name"])
user["name"] = "Ava"
print(user)
user["hobby"] = "Drawing"
print(user)

```

## 6) If statements
if-statements are amazing for letting the program do different things based on conditions.
If the condition is true the following code is acted out. 
If the condition is false the computer ignores the code and reads the next line of code instead. 

Use `if`, `elif`, `else` to choose what happens.

Try changing the score. Can you predict what code snippet runs?
Try changing the conditions.

Can you remember the difference between elif and else?



```python
score = 50
if score >= 90:
    print("A")
elif score >= 70:
    print("B")
else:
    print("C")
```

## 8) Functions
Functions let us name a block of code so we can reuse it.
This block of code does something, and that is why we call them functions. 

Overview:
    - We first define the function using the keyword def
    - Then you give it a name followed by parenthesis either with or without an argument.
    - Then you can call the function by using its name and give it an argument if needed in the parenthesis


```python
def greet(name):
    print("Hello, " + name + " !")

greet("Maya")
```
