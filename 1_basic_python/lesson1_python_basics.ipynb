## 1) Hello, Python (5 min)

## 2) Variables and types (6 min)
Python guesses the type based on the value.


```python
age = 15
name = "Ada"
height = 1.62
is_student = True

print(name, age, height, is_student)
print(type(age), type(name), type(height), type(is_student))
```

## 3) Strings (5 min)
Use `+` to combine variables and strings.


```python
my_name = "Your Name"
hobby = "your hobby"
print("Hi, I am "+ my_name + " and I like " + hobby + ".")
```

## 4) Lists (6 min)
Lists hold multiple items. Index starts at 0.


```python
colors = ["red", "green", "blue"]
print(colors)
print(colors[0])

colors.append("yellow")
print(colors)
```

## 5) Dictionaries (8 min)
Dictionaries store key-value pairs. This is similar to JSON used in web apps.


```python
user = {"id": 1, "name": "Maya", "age": 14}
print(user)

print(user["name"])
user["name"] = "Ava"
print(user)
user["hobby"] = "Drawing"
print(user)

```

## 6) If statements (5 min)
Use `if`, `elif`, `else` to choose what happens.


```python
score = 78
if score >= 90:
    print("A")
elif score >= 70:
    print("B")
else:
    print("C")
```

## 7) Loops (5 min)
Loops repeat actions. `for` loops are common with lists.


```python
animals = ["cat", "dog", "rabbit"]
for animal in animals:
    print(animal)
```

## 9) Mini challenge (5 min)
Make a `user` dictionary and a function that returns a greeting using the name inside the dictionary.


```python
user = {"id": 1, "name": "Your name"}

def greet_user(data):
    name = data["name"]
    return "Hi, " + name + "!"

print(greet_user(user))
```
