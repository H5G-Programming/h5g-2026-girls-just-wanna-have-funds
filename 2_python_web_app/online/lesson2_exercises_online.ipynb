Moving funds means: **subtract** from one goal, **add** to another.

Run the mini-example below first:

````python
# Mini-example: Moving money between two goals
sneakers = {"title": "Sneakers", "saved": 100}
art_set  = {"title": "Art set",  "saved": 30}

amount = 25  # we want to move 25 from sneakers to art_set

sneakers["saved"] = sneakers["saved"] - amount
art_set["saved"]  = art_set["saved"]  + amount

print("Sneakers:", sneakers)
print("Art set: ", art_set)
````

## Exercise 4: Close a Goal

**Concept:** Using an `if` statement to conditionally move money before closing a goal.

When you close a goal, any money you had saved doesn't just disappear!
If there's still money in it, it should move to another goal first.
But if the goal is already empty (saved amount is 0), there's nothing to move.

That's where an `if` statement comes in — it only runs code when a condition is true.

Run the mini-example below first:

````python
# Mini-example: Closing a goal and moving leftover funds
goal_closing = {"title": "Concert ticket", "saved": 40, "status": "active"}
goal_target  = {"title": "Sneakers",       "saved": 100}

amount = goal_closing["saved"]  # how much is left?

if amount > 0:  # only move money if there IS money to move
    goal_closing["saved"] = 0
    goal_target["saved"]  = goal_target["saved"] + amount

goal_closing["status"] = "closed"  # mark the goal as closed

print("Closed goal:", goal_closing)
print("Target goal:", goal_target)
````
