## 1) Link to the real app (5 min)
In `backend/app.py` there is a dictionary called `user` that looks like this:

```
user = {"id": 1, "name": "Maya"}
```

The `/api/user` endpoint returns this dictionary as JSON. If you change the `name`, the app will return something else.

### Challenge
Open `backend/app.py` and change the name to your own name. Save the file.
