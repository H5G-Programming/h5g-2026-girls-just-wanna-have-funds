## Exercise 6: Prioritize Your Goals

**Concept:** Using `range(len(...))` to loop through positions in a list.

The bank is called **Priority** Bank — so you should be able to drag your goals into the order that matters most to you!

When you drag a goal card to a new position, the frontend sends a list of goal IDs in the new order.
For example: `new_order = [3, 1, 2]` — this is a **list** of goal IDs.

Understanding `range(len(...))`:
- `new_order` is a list, e.g. `[3, 1, 2]`
- `len(new_order)` gives us how many items are in the list → `3`
- `range(3)` gives us the numbers `0, 1, 2` (it always starts at 0!)
- So `range(len(new_order))` gives us one number per item — perfect for positions!
- Inside the loop, `i` is the position number, and `new_order[i]` is the goal ID at that position.

Your job: loop through the list and set each goal's `"priority"` to its position number.

Run the mini-example below first to see how `range(len(...))` works:

````python
# Mini-example: Looping through positions in a list
colors = ["red", "blue", "green"]

for i in range(len(colors)):
    print("Position", i, "is", colors[i])
````

Expected output:
```
Position 0 is red
Position 1 is blue
Position 2 is green
```

See how `i` goes `0, 1, 2`? That's exactly what we need for priority numbers!
