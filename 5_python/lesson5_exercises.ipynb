## Exercise 6: Prioritize Your Goals

**Concept:** Using `range(len(...))` to loop through positions in a list.

The bank is called **Priority** Bank — so you should be able to drag your goals into the order that matters most to you!

When you drag a goal card to a new position, the frontend sends a list of goal IDs in the new order.
For example: `new_order = [3, 1, 2]` — this is a **list** of goal IDs.

Understanding `range(len(...))`:
- `new_order` is a list, e.g. `[3, 1, 2]`
- `len(new_order)` gives us how many items are in the list → `3`
- `range(3)` gives us the numbers `0, 1, 2` (it always starts at 0!)
- So `range(len(new_order))` gives us one number per item — perfect for positions!
- Inside the loop, `i` is the position number, and `new_order[i]` is the goal ID at that position.

Your job: loop through the list and set each goal's `"priority"` to its position number.

Run the mini-example below first to see how `range(len(...))` works:

````python
# Mini-example: Looping through positions in a list
colors = ["red", "blue", "green"]

for i in range(len(colors)):
    print("Position", i, "is", colors[i])
````

Expected output:
```
Position 0 is red
Position 1 is blue
Position 2 is green
```

See how `i` goes `0, 1, 2`? That's exactly what we need for priority numbers!

## Exercise 7: Find Your Closest Goal

**Concept:** The "best so far" pattern — keeping track of the best value while looping.

The summary page shows which goal you're closest to reaching.
Right now, the code uses a Python shortcut (`max(...)`) to find it — but we want to write the loop ourselves!

The idea is simple:
- Start with `best_percentage = 0` (nothing found yet)
- Loop through every goal
- For each goal, calculate how close you are: `saved_amount / target_amount`
- If this goal's percentage is **better** than what we've seen so far, remember it!

This is called the **"best so far"** pattern — you keep updating your answer as you go.

Run the mini-example below to see the pattern in action:

````python
# Mini-example: Find the tallest height from a list
heights = [165, 172, 158, 180, 169]

tallest = 0

for i in range(len(heights)):
    if heights[i] > tallest:
        tallest = heights[i]

print("The tallest person is", tallest, "cm")
````

Expected output:
```
The tallest person is 180 cm
```

See the pattern? We start with `tallest = 0`, and every time we find something bigger, we update it. At the end, `tallest` holds the best value we found!

## Exercise 8: Smart Emoji Picker

**Concept:** String matching with the `in` keyword + `if/elif/else`.

When you create a new goal, the app always uses 💜 as the default emoji. Boring!

Let's make it smarter — if someone creates a goal called "New sneakers", the app should automatically pick 👟!

The trick is Python's `in` keyword. It checks if a small string appears inside a bigger string:

```python
"shoe" in "new shoes for school"   # True ✅
"pizza" in "save for concert"      # False ❌
```

We also use `.lower()` to make it case-insensitive (so "SHOES" and "shoes" both match).

Run the mini-example below:

````python
# Mini-example: Pick an emoji based on keywords in a message
message = "I love playing football"
message_lower = message.lower()

if "football" in message_lower or "soccer" in message_lower:
    emoji = "⚽"
elif "music" in message_lower or "sing" in message_lower:
    emoji = "🎵"
else:
    emoji = "😊"

print(emoji, message)
````

Expected output:
```
⚽ I love playing football
```

Try changing the `message` to "I love singing" or "hello world" and run it again — does the emoji change?
